## **2.- Análisis Descriptivo y Visualización**
Calcula estadísticas descriptivas básicas para comprender mejor las características de los datos.

````python
# Convertir los campos tipo textos a fechas
df['Inward Date'] = pd.to_datetime(df['Inward Date'], errors='coerce')
df['Dispatch Date'] = pd.to_datetime(df['Dispatch Date'], errors='coerce')

# Para el analisis por categioras falta la ventas totales
df['Venta Total'] = df['Price'] * df['Quantity Sold']


print('Estadisticas de lso camppos cantidad, precio y vental total')
print(df[['Price', 'Quantity Sold', 'Venta Total']].describe())

print('--'*80)
print('Venta por producto')
print(df.groupby('Product')['Venta Total'].sum().sort_values(ascending=False))

print('--'*80)
print('Venta por Marca')
print(df.groupby('Brand')['Venta Total'].sum().sort_values(ascending=False).head(10))

print('--'*80)
print('Venta por Region')
print(df.groupby('Region')['Venta Total'].sum().sort_values(ascending=False))
````

Utiliza histogramas y boxplots para visualizar la distribución de las variables numéricas clave.

````python
import matplotlib.pyplot as plt

# Histograma para los precios
plt.figure(figsize=(12, 6))
plt.hist(df['Price'], bins=30, edgecolor='blue', alpha=0.4)
plt.title('Histograma de Precios')
plt.xlabel('Precio')
plt.ylabel('Cantidad')
plt.grid(True)
plt.show()

# Histograma para las cantidad de unidades
plt.figure(figsize=(12, 6))
plt.hist(df['Quantity Sold'], bins=30, edgecolor='coral', alpha=0.4)
plt.title('Histograma de Unidades Vendidas')
plt.xlabel('Cantidad Vendida')
plt.ylabel('Cantidad')
plt.grid(True)
plt.show()

# Histograma para Ventas Totales
plt.figure(figsize=(8, 5))
plt.hist(df['Venta Total'], bins=30, edgecolor='red', alpha=0.7)
plt.title('Histograma de Ventas Totales')
plt.xlabel('Ventas Totales')
plt.ylabel('Cantidad')
plt.grid(True)
plt.show()
````

Analisis del grafico 1 apunta que todos los producto con distintos precios tuvieron similares venta de unidades

Analisis del grafico 2 como el grafico anterior la cantidad vendida por producto fue muy similar en el rango de los 5000

Analisis del grafico 3 hay un sesgo a la derech, producto de pocas ventas generan venta de valores muy altos

````python


# Boxplot de Precios
plt.figure(figsize=(12, 6))
plt.boxplot(df['Price'], vert=False)
plt.title('Boxplot de Precios')
plt.xlabel('Precio')
plt.grid(True)
plt.show()


# Boxplot de unidades vendidas
plt.figure(figsize=(12, 6))
plt.boxplot(df['Quantity Sold'], vert=False)
plt.title('Boxplot de Unidades Vendidas')
plt.xlabel('Cantidad Vendida')
plt.grid(True)
plt.show()

# Boxplot de Ventas totales
plt.figure(figsize=(12, 6))
plt.boxplot(df['Venta Total'], vert=False)
plt.title('Boxplot de Ventas Totales')
plt.xlabel('Ventas Totales')
plt.grid(True)
plt.show()
````

El analisis de  Boxplot de Precios indica que los valores mas altos pueden ser laptop o celurares de alta gama

El analisis de  Boxplot de Unidades para altas cantidades puede deberse a Promociones o alguna venta por algun tipo de convenio

El analisis de  Boxplot de Ventas totales puede deberse a los analisis anteriores equipo de lata gama y promociones por volumen.

**Crea gráficos de líneas para mostrar las tendencias de ventas a lo largo del tiempo.**

````python
# Agrupar las ventas por fecha ordenadas por el campo fecha de Despacho
ventas_tiempo = (df.groupby('Dispatch Date')['Venta Total'].sum().reset_index().sort_values('Dispatch Date'))

plt.figure(figsize=(12, 6))
plt.plot(ventas_tiempo['Dispatch Date'], ventas_tiempo['Venta Total'], marker='o')
plt.title('Tendencia de Ventas Totales en el periodo de venta')
plt.xlabel('Fecha de Despacho')
plt.ylabel('Ventas Totales')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
````

Utiliza gráficos de dispersión para analizar la relación entre diferentes variables.

````python
# Analisis de Meses por año
df['Mes'] = df['Dispatch Date'].dt.to_period('M').astype(str)

ventas_mensuales = df.groupby('Mes')['Venta Total'].sum().reset_index()

plt.figure(figsize=(12, 5))
plt.plot(ventas_mensuales['Mes'], ventas_mensuales['Venta Total'], marker='o')
plt.title('Ventas Mensuales por Año')
plt.xlabel('Mes')
plt.ylabel('Ventas Totales')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
````

````python
# Analisis de Precio en Venta Total

plt.figure(figsize=(8, 5))
plt.scatter(df['Price'], df['Venta Total'], alpha=0.5)
plt.title('Relación entre Precio y Ventas Totales')
plt.xlabel('Precio')
plt.ylabel('Ventas Totales')
plt.grid(True)
plt.show()

````

Confima que a precios mas alto mayores ventas

**Combina histogramas y boxplots para visualizar la distribución de las ventas y detectar valores atípicos.**

````python
# Analisis de las ventas totales

fig, axes = plt.subplots(2, 1, figsize=(10, 7), gridspec_kw={'height_ratios': [3, 1]})

axes[0].hist(df['Venta Total'], bins=30, edgecolor='black', alpha=0.7)
axes[0].set_title('Distribución de Ventas Totales')
axes[0].set_xlabel('Ventas Totales')
axes[0].set_ylabel('Frecuencia')
axes[0].grid(True)

# boxplots Horizontal para revisar valores atipicos
axes[1].boxplot(df['Venta Total'], vert=False)
axes[1].set_title('Boxplot de Ventas Totales')
axes[1].set_xlabel('Ventas Totales')
axes[1].grid(True)

plt.tight_layout()
plt.show()
````

El analisis permite visualizar la concetracion de ventas y la dispersion de los datos y los valores tipicos en los extremos como equipo de alta gama que se alejan del promedio
